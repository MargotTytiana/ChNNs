```python
from chaotic_network import ChaoticSpeakerRecognitionNetwork
import torch

config = {
    'sample_rate': 16000,
    'embedding_dim': 10,
    'mlsa_scales': 5,
    'num_speakers': 251,
    'device': 'cpu'
}

model = ChaoticSpeakerRecognitionNetwork(**config)
print(f'use_differentiable_features: {model.use_differentiable_features}')

# 测试梯度流
audio = torch.randn(2, 8000, requires_grad=True)
output = model(audio)
output.sum().backward()
print(f'Input gradient exists: {audio.grad is not None}')
```
