### 在配置文件中添加：

```yaml
loss_config:
  adversarial:
    enabled: true                 # 关键！必须为 true
    perturbation_scale: 0.1       # 扰动尺度
    parameter_noise: true         # 参数扰动
    trajectory_noise: true        # 轨迹扰动
    warmup_epochs: 5              # 预热期
    adversarial_weight: 0.3       # 损失权重 α
```

---

## 对比实验建议

如果你想验证对抗训练的效果，建议进行以下对比：

| 实验组 | 配置 | 目的 |
|--------|------|------|
| **Baseline** | `enabled: false` | 无对抗训练 |
| **Adv-0.1** | `enabled: true, perturbation_scale: 0.1` | 默认扰动 |
| **Adv-0.05** | `enabled: true, perturbation_scale: 0.05` | 较小扰动 |
| **Adv-0.2** | `enabled: true, perturbation_scale: 0.2` | 较大扰动 |

### 评估指标

1. **Clean Accuracy**：无噪声条件下的准确率
2. **Noisy Accuracy**：添加噪声后的准确率（SNR = 10dB, 20dB）
3. **Performance Retention**：`Noisy Acc / Clean Acc`
